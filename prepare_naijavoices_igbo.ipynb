# Prepare NaijaVoices dataset

```python
train_path = "../naijavoices/train.csv"
dev_path = "../naijavoices/dev.csv"
test_path = "../naijavoices/test.csv"

train_df = pd.read_csv(train_path)
dev_df = pd.read_csv(dev_path)
test_df = pd.read_csv(test_path)
```

```python
# get Igbo subset
train_df = train_df[train_df['language'] == 'igbo']
dev_df = dev_df[dev_df['language'] == 'igbo']
test_df = test_df[test_df['language'] == 'igbo']

# drop language column
train_df = train_df.drop(columns=['language'])
dev_df = dev_df.drop(columns=['language'])
test_df = test_df.drop(columns=['language'])
```

```python
# drop unnecessary columns
drop_cols = ['phase', 'exists', 'snr', 'speaker_id']
train_df.drop(columns=drop_cols + ['gender', 'age_bracket'], inplace=True)
dev_df.drop(columns=drop_cols + ['gender', 'age_bracket'], inplace=True)
test_df.drop(columns=drop_cols, inplace=True)
```

```python
# drop duplicates
train_df = train_df.drop_duplicates(subset='audio')
dev_df = dev_df.drop_duplicates(subset='audio')
test_df = test_df.drop_duplicates(subset='audio')
```

```python
# get the audio file paths
path = '../naijavoices/'

train_df['audio_filepath'] = train_df['audio'].apply(lambda x: os.path.join(os.path.dirname(path), "audio-unconverted", x))
dev_df['audio_filepath'] = dev_df['audio'].apply(lambda x: os.path.join(os.path.dirname(path), "audio-unconverted", x))
test_df['audio_filepath'] = test_df['audio'].apply(lambda x: os.path.join(os.path.dirname(path), "audio-unconverted", x))

# drop the `audio` column
train_df.drop(columns=['audio'], inplace=True)
dev_df.drop(columns=['audio'], inplace=True)
test_df.drop(columns=['audio'], inplace=True)
```

```python
# remove long audio files from train and dev sets
train_df = train_df[train_df['duration'] <= 15]
dev_df = dev_df[dev_df['duration'] <= 15]
print(f"Total training set duration: {train_df['duration'].sum() / 3600:.2f} hours")
```

```python
train_df.head()
```

```python
# save original text before cleaning
train_df['original_text'] = train_df['text']
dev_df['original_text'] = dev_df['text']
test_df['original_text'] = test_df['text']
```

```python
# lowercase the text
train_df['text'] = train_df['text'].str.lower()
dev_df['text'] = dev_df['text'].str.lower()
test_df['text'] = test_df['text'].str.lower()

# get the set of characters in the text
train_text = "".join(train_df['text'].tolist())
dev_text = "".join(dev_df['text'].tolist())
text = train_text + dev_text

set(text)
```

```python
char = '\u200b'

# find rows with the character
train_df[train_df['text'].str.contains(char)]
```

```python
print(f"'{char}'")
```

```python
# remove punctuation and special characters
CHARS_TO_REMOVE = ['"', '!', 'ǃ', '(', ')', ',', '.', ':', ';', '?', '[', ']', '\u200b']

for char in CHARS_TO_REMOVE:
    train_df['text'] = train_df['text'].str.replace(char, '', regex=False)
    dev_df['text'] = dev_df['text'].str.replace(char, '', regex=False)
    test_df['text'] = test_df['text'].str.replace(char, '', regex=False)

# replace whitespace characters with a single space
WHITESPACE_CHARS = ['\xa0', '\t', '/']
for char in WHITESPACE_CHARS:
    train_df['text'] = train_df['text'].str.replace(char, ' ', regex=False)
    dev_df['text'] = dev_df['text'].str.replace(char, ' ', regex=False)
    test_df['text'] = test_df['text'].str.replace(char, ' ', regex=False)

# strip leading and trailing whitespace from the text column
train_df['text'] = train_df['text'].str.strip()
dev_df['text'] = dev_df['text'].str.strip()
test_df['text'] = test_df['text'].str.strip()

# remove excess whitespace
train_df['text'] = train_df['text'].str.replace(r'\s+', ' ', regex=True)
dev_df['text'] = dev_df['text'].str.replace(r'\s+', ' ', regex=True)
test_df['text'] = test_df['text'].str.replace(r'\s+', ' ', regex=True)
```

```python
# get the frequency of each character in the train and dev sets
char_freq = Counter(text)
char_freq
```

```python
# remove rows with infrequent characters from train and dev sets
min_freq = 95
train_df = train_df[train_df['text'].apply(lambda x: all(char_freq[c] >= min_freq for c in x))]
dev_df = dev_df[dev_df['text'].apply(lambda x: all(char_freq[c] >= min_freq for c in x))]
```

```python
# recalculate character frequency after cleaning
train_text = "".join(train_df['text'].tolist())
dev_text = "".join(dev_df['text'].tolist())
text = train_text + dev_text

# get the frequency of each character in the train and dev sets
char_freq = Counter(text)
char_freq
```

```python
# remove rows with anomalous character rates
train_df['character_rate'] = train_df['text'].str.len() / train_df['duration']
train_df['character_rate'].describe()
```

```python
dev_df['character_rate'] = dev_df['text'].str.len() / dev_df['duration']
dev_df['character_rate'].describe()
```

```python
MAX = 23
train_df = train_df[train_df['character_rate'] <= MAX]
dev_df = dev_df[dev_df['character_rate'] <= MAX]

train_df.drop(columns=['character_rate'], inplace=True)
dev_df.drop(columns=['character_rate'], inplace=True)
```

```python
# save to NeMo jsonl format
train_manifest_path = os.path.join(os.path.dirname(path), "igbo_train_manifest.jsonl")
dev_manifest_path = os.path.join(os.path.dirname(path), "igbo_dev_manifest.jsonl")
test_manifest_path = os.path.join(os.path.dirname(path), "igbo_test_manifest.jsonl")

train_df.to_json(train_manifest_path, orient='records', lines=True)
dev_df.to_json(dev_manifest_path, orient='records', lines=True)
test_df.to_json(test_manifest_path, orient='records', lines=True)
print(f"Manifests saved to {path}")
```

```python
from datasets import Dataset, DatasetDict, Audio

dev = Dataset.from_pandas(dev_df)
test = Dataset.from_pandas(test_df)
```

```python
# only keep necessary columns
cols = ['audio', 'text', 'duration']
dev = dev.remove_columns([col for col in dev.column_names if col not in cols])
test = test.remove_columns([col for col in test.column_names if col not in cols + ['gender', 'age_bracket']])
```

```python
# cast audio column to Audio type
dev = dev.cast_column("audio", Audio(16000))
test = test.cast_column("audio", Audio(16000))
```

```python
def sample_dataset(df, duration_minutes, manifest_name, random_state=1):
    """
    Sample a dataset to a specific duration.
    """
    total_duration = float(df['duration'].sum())
    ratio = (duration_minutes * 60) / total_duration
    n = int(len(df) * ratio)
    sample = df.sample(n=n, random_state=random_state)
    sample = sample.sample(frac=1, random_state=random_state).reset_index(drop=True)
    sample_duration = sample['duration'].sum()
    if sample_duration >= 3600:
        print(f"Sample duration: {sample_duration / 3600:.2f} hours")
    else:
        print(f"Sample duration: {sample_duration / 60:.2f} minutes")

    # save to new manifest
    sample_manifest_path = os.path.join(
        os.path.dirname(path),
        f"igbo_{manifest_name}_{int(sample_duration / 3600)}h.jsonl"
    )
    sample.to_json(sample_manifest_path, orient="records", lines=True)
    print(f"Manifest saved to {sample_manifest_path}")

    dataset = Dataset.from_pandas(sample)
    dataset = dataset.remove_columns([col for col in dataset.column_names if col not in cols])
    dataset = dataset.cast_column("audio", Audio(16000))
    return dataset
```

```python
train = sample_dataset(train_df, 500 * 60, "train_manifest", random_state=10)

# save to disk
dataset = DatasetDict({
    'train': train,
    'validation': dev,
    'test': test
})
# dataset.save_to_disk("naijavoices_igbo_500h")
```

```python
from IPython.display import Audio as DisplayAudio
import random

def display_audio_sample(dataset, index):
    """
    Display an audio sample from the dataset.
    """
    audio = dataset[index]['audio']
    text = dataset[index]['text']
    print(f"Text: {text}")
    print(f"Duration: {dataset[index]['duration']} seconds")
    print(f"Sampling rate: {audio['sampling_rate']} Hz")
    return DisplayAudio(audio['array'], rate=audio['sampling_rate'])
```
